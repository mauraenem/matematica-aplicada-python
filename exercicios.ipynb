```python
from numpy import pi
from math import factorial
x_graus=float(input('Informe o valor do ângulo, em graus: '))
termos=int(input('Informe o número de termos do somatório: '))
x_radianos=pi*x_graus/180
seno=0
for n in range (termos):
  seno=seno+(-1)**n*x_radianos**(2*n+1)/factorial(2*n+1)
print(seno)
```

```python
import pandas as pd
x={'Pesos':[48.8, 48.9, 49.0, 49.1, 49.2, 49.3, 49.5, 49.7, 49.7,
            49.7, 49.8, 50.0, 50.1, 50.1, 50.2, 50.2, 50.4, 50.6,
            50.8, 50.9]}
```

```python
p=pd.DataFrame(x)
media=p['Pesos'].mean()
moda=p['Pesos'].mode()
mediana=p['Pesos'].median()
print(f'Média: {media}')
print(f'Moda: {moda}')
print(f'Mediana: {mediana}')
```
